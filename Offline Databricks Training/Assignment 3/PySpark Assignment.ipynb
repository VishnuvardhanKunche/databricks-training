**Part A DataFrame Creation**

- Create the Employee DataFrame using a list of tuples.
- Create the Department DataFrame.
- Create the Manager DataFrame.
- Display the schema of all DataFrames.
- Display the first 5 rows.

```python
employee_data = [
    (101, "Alice", 25, "HR", 45000, "Chennai", "2022-01-15", 201),
    (102, "Bob", 30, "IT", 70000, "Bangalore", "2021-06-20", 202),
    (103, "Charlie", None, "IT", None, "Chennai", "2023-03-12", 202),
    (104, "David", 28, "Finance", 65000, "Mumbai", "2020-09-18", 203),
    (105, "Eva", 35, "HR", 80000, None, "2019-05-25", 201),
    (106, "Frank", 29, "Marketing", 55000, "Hyderabad", None, 204),
    (107, "Grace", 31, "Finance", None, "Pune", "2022-12-01", 203),
    (108, "Henry", 26, "IT", 60000, "Bangalore", "2024-01-10", None)
]

employee_columns = [
    "emp_id",
    "name",
    "age",
    "department",
    "salary",
    "city",
    "joining_date",
    "manager_id"
]

employee_df = spark.createDataFrame(employee_data, employee_columns)

employee_df.printSchema()
employee_df.show(5)

department_data = [
    ("HR", "Chennai"),
    ("IT", "Bangalore"),
    ("Finance", "Mumbai"),
    ("Marketing", "Hyderabad"),
    ("Sales", "Delhi")
]

department_columns = [
    "dept_name",
    "location"
]

department_df = spark.createDataFrame(department_data, department_columns)

department_df.printSchema()
department_df.show(5)


manager_data = [
    (201, "Robert"),
    (202, "Jennifer"),
    (203, "Michael"),
    (204, "Sophia")
]

manager_columns = [
    "manager_id",
    "manager_name"
]

manager_df = spark.createDataFrame(manager_data, manager_columns)

manager_df.printSchema()
manager_df.show(5)
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import *
from pyspark.sql.window import Window
```

**Part B  Select Operations**
- Select only emp_id, name, and salary.
- Select all employees from the IT department.
- Select employee names and alias the column as Employee_Name.
- Select salary after increasing it by 10%.
- Select employees whose city is Chennai or Bangalore.

```python
# 6. Select emp_id, name, salary

employee_df.select(
    "emp_id",
    "name",
    "salary"
).show()


# 7. Employees from IT department

employee_df.filter(
    F.col("department") == "IT"
).show()


# 8. Alias employee name

employee_df.select(
    F.col("name").alias("Employee_Name")
).show()


# 9. Increase salary by 10%

employee_df.select(
    "emp_id",
    "name",
    "salary",
    (F.col("salary") * 1.10).alias("updated_salary")
).show()


# 10. Employees from Chennai or Bangalore

employee_df.filter(
    F.col("city").isin("Chennai", "Bangalore")
).show()
```

**Part C - withColumn()**
- Create a new column called bonus equal to 20% of salary.
- Create a column called annual_salary.
- Create a column called experience_bonus where:
    • Salary > 70000 → 10000
    • Salary between 50000 and 70000 → 5000
    • Otherwise → 2000
- Convert employee names to uppercase.
- Add a column called country with value "India".


```python
# 11. Bonus = 20% of salary

employee_df.withColumn(
    "bonus",
    F.col("salary") * 0.20
).show()      # Output 16


# 12. Annual Salary

employee_df.withColumn(
    "annual_salary",
    F.col("salary") * 12
).show()      # Output 17


# 13. Experience Bonus

employee_df.withColumn(
    "experience_bonus",
    F.when(F.col("salary") > 70000, 10000)
     .when((F.col("salary") >= 50000) & (F.col("salary") <= 70000), 5000)
     .otherwise(2000)
).show()      # Output 18


# 14. Convert names to uppercase

employee_df.withColumn(
    "name",
    upper("name")
).show()      # Output 19


# 15. Add country column

employee_df.withColumn(
    "country",
    F.lit("India")
).show()      # Output 20
```

**Part D - withColumnRenamed()**

- Rename multiple columns one by one.

```python
# 16. Rename department

employee_df.withColumnRenamed(
    "department",
    "dept"
).show()      # Output 21


# 17. Rename salary

employee_df.withColumnRenamed(
    "salary",
    "monthly_salary"
).show()      # Output 22


# 18. Rename multiple columns

employee_df.withColumnRenamed(
    "department",
    "dept"
).withColumnRenamed(
    "salary",
    "monthly_salary"
).withColumnRenamed(
    "city",
    "location"
).show()      # Output 23
```

- Perform a full outer join.

```python
# 49. Inner Join

employee_df.join(
    department_df,
    employee_df.department == department_df.dept_name,
    "inner"
).show()      # Output 54


# 50. Left Join

employee_df.join(
    department_df,
    employee_df.department == department_df.dept_name,
    "left"
).show()      # Output 55


# 51. Right Join

employee_df.join(
    department_df,
    employee_df.department == department_df.dept_name,
    "right"
).show()      # Output 56


# 52. Full Outer Join

employee_df.join(
    department_df,
    employee_df.department == department_df.dept_name,
    "outer"
).show()      # Output 57


# 53. Join Employee with Manager

employee_df.join(
    manager_df,
    on="manager_id",
    how="left"
).show()      # Output 58


# 54. Employee Name and Manager Name

employee_df.join(
    manager_df,
    on="manager_id",
    how="left"
).select(
    "name",
    "manager_name"
).show()      # Output 59


# 55. Employees whose city matches department location

employee_df.join(
    department_df,
    employee_df.department == department_df.dept_name,
    "inner"
).filter(
    employee_df.city == department_df.location
).select(
    "emp_id",
    "name",
    "department",
    "city"
).show()      # Output 60


# 56. Departments having no employees

department_df.join(
    employee_df,
    department_df.dept_name == employee_df.department,
    "left_anti"
).show()      # Output 61
```

**Part N - Advanced Questions**
- Display employees whose salary is above the department average.
- Find the second highest salary.
- Find duplicate cities.
- Display employees who joined in the last two years.
- Replace NULL manager IDs with 999.
- Create a salary grade:
 •	A → Salary ≥ 80000
 •	B → Salary ≥ 60000
 •	C → Salary ≥ 40000
 •	D → Otherwise
- Create an employee ID string like EMP101.
- Find employees whose names end with "e".
- Display employees sorted by joining date.
- Find employees who have worked for more than three years.


```python
# 81. Employees whose salary is above department average

dept_avg = employee_df.groupBy("department") \
    .agg(avg("salary").alias("avg_salary"))

employee_df.alias("e") \
.join(
    dept_avg.alias("d"),
    "department"
).filter(
    col("salary") > col("avg_salary")
).select(
    "emp_id",
    "name",
    "department",
    "salary",
    "avg_salary"
).show()      # Output 86


# 82. Second highest salary

employee_df.select("salary") \
.distinct() \
.orderBy(col("salary").desc()) \
.limit(2) \
.orderBy("salary") \
.show()      # Output 87


# 83. Duplicate cities

employee_df.groupBy("city") \
.count() \
.filter(col("count") > 1) \
.show()      # Output 88


# 84. Employees joined in the last two years

employee_date_df.filter(
    year(current_date()) - year(col("joining_date")) <= 2
).show()      # Output 89


# 85. Replace NULL manager IDs with 999

employee_df.fillna(
    {"manager_id":999}
).show()      # Output 90


# 86. Salary Grade

employee_df.withColumn(
    "salary_grade",
    when(col("salary") >= 80000,"A")
    .when(col("salary") >= 60000,"B")
    .when(col("salary") >= 40000,"C")
    .otherwise("D")
).show()      # Output 91


# 87. Employee ID String

employee_df.withColumn(
    "employee_code",
    concat(lit("EMP"), col("emp_id"))
).show()      # Output 92


# 88. Employee names ending with 'e'

employee_df.filter(
    col("name").endswith("e")
).show()      # Output 93


# 89. Employees sorted by joining date

employee_date_df.orderBy(
    "joining_date"
).show()      # Output 94


# 90. Employees worked for more than 3 years

employee_date_df.filter(
    (datediff(current_date(),"joining_date")/365) > 3
).show()      # Output 95
```

**Challenge Questions**
- Rank employees by salary within each department.
- Find the top 2 highest-paid employees in every department.
- Find departments with average salary greater than 60000.
- Find employees who do not have a manager.
- Display employees with manager names and department locations in a single DataFrame.
- Identify employees with missing information (NULL values).
- Replace missing salaries with the department average salary.
- Calculate monthly tax as 10% of salary.
- Find employees who joined on weekends.
- Create a final report containing:
•	Employee Name
•	Department
•	Manager Name
•	Department Location
•	Salary
•	Bonus
•	Annual Salary
•	Years Worked
•	Salary Grade



```python
# 91. Rank employees by salary within each department

window_spec = Window.partitionBy("department").orderBy(col("salary").desc())

employee_df.withColumn(
    "salary_rank",
    rank().over(window_spec)
).show()      # Output 96


# 92. Top 2 highest-paid employees in every department

employee_df.withColumn(
    "row_num",
    row_number().over(window_spec)
).filter(
    col("row_num") <= 2
).drop(
    "row_num"
).show()      # Output 97


# 93. Departments with average salary greater than 60000

employee_df.groupBy(
    "department"
).agg(
    avg("salary").alias("average_salary")
).filter(
    col("average_salary") > 60000
).show()      # Output 98


# 94. Employees who do not have a manager

employee_df.filter(
    col("manager_id").isNull()
).show()      # Output 99


# 95. Employee with manager name and department location

employee_df.alias("e") \
.join(
    manager_df.alias("m"),
    "manager_id",
    "left"
).join(
    department_df.alias("d"),
    col("e.department") == col("d.dept_name"),
    "left"
).select(
    col("e.emp_id"),
    col("e.name"),
    col("m.manager_name"),
    col("d.location")
).show()      # Output 100


# 96. Employees with missing information

employee_df.filter(
    col("age").isNull() |
    col("salary").isNull() |
    col("city").isNull() |
    col("joining_date").isNull() |
    col("manager_id").isNull()
).show()      # Output 101


# 97. Replace NULL salary with department average salary

department_avg = employee_df.groupBy("department") \
    .agg(avg("salary").alias("department_average_salary"))

employee_df.alias("e") \
.join(
    department_avg.alias("d"),
    "department"
).withColumn(
    "updated_salary",
    when(
        col("salary").isNull(),
        col("department_average_salary")
    ).otherwise(col("salary"))
).drop(
    "department_average_salary"
).show()      # Output 102


# 98. Monthly tax as 10% of salary

employee_df.withColumn(
    "monthly_tax",
    round(col("salary") * 0.10, 2)
).show()      # Output 103


# 99. Employees who joined on weekends

employee_date_df.filter(
    dayofweek("joining_date").isin([1, 7])
).show()      # Output 104


# 100. Final Employee Report

employee_df.alias("e") \
.join(
    manager_df.alias("m"),
    "manager_id",
    "left"
).join(
    department_df.alias("d"),
    col("e.department") == col("d.dept_name"),
    "left"
).withColumn(
    "bonus",
    col("salary") * 0.20
).withColumn(
    "annual_salary",
    col("salary") * 12
).withColumn(
    "years_worked",
    round(datediff(current_date(), to_date("joining_date")) / 365, 2)
).withColumn(
    "salary_grade",
    when(col("salary") >= 80000, "A")
    .when(col("salary") >= 60000, "B")
    .when(col("salary") >= 40000, "C")
    .otherwise("D")
).select(
    col("e.name").alias("Employee Name"),
    col("e.department").alias("Department"),
    col("m.manager_name").alias("Manager Name"),
    col("d.location").alias("Department Location"),
    col("e.salary").alias("Salary"),
    col("bonus").alias("Bonus"),
    col("annual_salary").alias("Annual Salary"),
    col("years_worked").alias("Years Worked"),
    col("salary_grade").alias("Salary Grade")
).show()      # Output 105
```
